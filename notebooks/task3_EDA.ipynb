```python
# Import Visualization Libraries
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
# Load Cleaned Dataset
import pandas as pd
df = pd.read_csv("cleaned_titanic.csv")
```

```python
# Basic Statistical Summary
df.describe()
```

```python
# Survival Count Analysis
df['Survived'].value_counts()
```

```python
# Visualize Survival Count
sns.countplot(x='Survived', data=df)

plt.title("Survival Count")
plt.show()
```

```python
# Gender vs Survival
sns.countplot(x='Sex', hue='Survived', data=df)

plt.title("Survival by Gender")
plt.show()
```

```python
# Passenger Class Analysis
sns.countplot(x='Pclass', hue='Survived', data=df)

plt.title("Passenger Class vs Survival")
plt.show()
```

```python
# Age Distribution Histogram
sns.histplot(df['Age'], bins=20)

plt.title("Age Distribution")
plt.show()
```

```python
# Correlation Heatmap
correlation = df.corr(numeric_only=True)

sns.heatmap(correlation, annot=True)

plt.title("Correlation Heatmap")
plt.show()
```

```python
# Detect Outliers
sns.boxplot(x=df['Fare'])

plt.title("Fare Outliers")
plt.show()
```

# Exploratory Data Analysis Findings

## Key Insights

1. More passengers died than survived.
2. Female passengers had higher survival rates.
3. First-class passengers survived more frequently.
4. Most passengers were between ages 20–40.
5. Fare column contains some outliers.
6. Passenger class has a relationship with survival rate.
