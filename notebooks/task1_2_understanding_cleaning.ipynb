TASK 2: 
Data Cleaning & Preprocessing 

Handle missing values
Remove duplicates
Clean dataset professionally

```python
# Check Missing Values

df.isnull().sum()
```

```python
# Fill Missing Age Values
# Best beginner method:Fill missing ages with average age.

df['Age'] = df['Age'].fillna(df['Age'].mean())
```

```python
df.isnull().sum()
```

```python
# Handle Cabin Column
# Cabin has too many missing values - Drop the column completely

df.drop('Cabin', axis=1, inplace=True)
```

```python
df.isnull().sum()
```

```python
# Handle Embarked Missing Values
# Embarked is categorical/text data - Fill with most common value.

df['Embarked'] = df['Embarked'].fillna(df['Embarked'].mode()[0])
```

```python
df.isnull().sum()
```

```python
# Check Duplicate Rows
df.duplicated().sum()
```

```python
# Remove Duplicates
df.drop_duplicates(inplace=True)
```

```python
# Check Dataset Shape Again
df.shape
```

```python
# Save Cleaned Dataset
df.to_csv("cleaned_titanic.csv", index=False)
```

# Data Cleaning & Preprocessing

The dataset contained missing values in Age, Cabin, and Embarked columns.

Cleaning steps performed:
- Filled missing Age values using mean
- Dropped Cabin column due to excessive missing values
- Filled Embarked missing values using mode
- Removed duplicate rows
- Saved cleaned dataset
