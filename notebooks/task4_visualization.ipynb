```python
# Import Libraries
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
# Load Cleaned Dataset
df = pd.read_csv("cleaned_titanic.csv")
```

```python
# Set Better Visualization Style
sns.set_style("whitegrid")
```

```python
# VISUALIZATION 1 — Survival Count Chart
plt.figure(figsize=(6,4))

sns.countplot(x='Survived', data=df)

plt.title("Survival Count")
plt.xlabel("Survived")
plt.ylabel("Passenger Count")

plt.show()
```

```python
# VISUALIZATION 2 — Gender Survival Analysis
plt.figure(figsize=(7,5))

sns.countplot(x='Sex', hue='Survived', data=df)

plt.title("Survival by Gender")

plt.show()
```

```python
# VISUALIZATION 3 — Passenger Class Distribution
plt.figure(figsize=(7,5))

sns.countplot(x='Pclass', data=df)

plt.title("Passenger Class Distribution")

plt.show()
```

```python
# VISUALIZATION 4 — Age Distribution Histogram
plt.figure(figsize=(8,5))

sns.histplot(df['Age'], bins=20)

plt.title("Age Distribution")

plt.show()
```

```python
# VISUALIZATION 5 — Fare Distribution
plt.figure(figsize=(8,5))

sns.histplot(df['Fare'], bins=30)

plt.title("Fare Distribution")

plt.show()
```

```python
# VISUALIZATION 6 — Correlation Heatmap
plt.figure(figsize=(10,6))

correlation = df.corr(numeric_only=True)

sns.heatmap(correlation, annot=True)

plt.title("Correlation Heatmap")

plt.show()
```

```python
# VISUALIZATION 7 — Boxplot for Outliers
plt.figure(figsize=(8,4))

sns.boxplot(x=df['Fare'])

plt.title("Fare Outliers")

plt.show()
```

```python
# Save Graph Images
plt.figure(figsize=(6,4))

sns.countplot(x='Survived', data=df)

plt.title("Survival Count")

plt.savefig("survival_count.png")

plt.show()
```

# Data Visualization

Different visualizations were created to analyze passenger survival patterns, age distribution, fare distribution, and relationships between variables.

Visualizations included:
- Count plots
- Histograms
- Heatmaps
- Boxplots

These charts helped identify trends, correlations, and outliers in the dataset.
